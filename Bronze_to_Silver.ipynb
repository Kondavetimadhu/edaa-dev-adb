```python
dbutils.widgets.text("landing_timestamp", "")
dbutils.widgets.text("table_name", "")

landing_timestamp = dbutils.widgets.get("landing_timestamp")
table_name = dbutils.widgets.get("table_name")
```

```python
bronze_df = spark.read.table(f"adb_edaa_dev_ind.bronze.{table_name}").filter(f"landing_timestamp = '{landing_timestamp}'")

```

```python
from pyspark.sql import functions as F
updated_df = bronze_df.withColumn("insert_timestamp", F.current_timestamp()).withColumn("updated_timestamp", F.current_timestamp())
```

```python
updated_df.write.mode("overwrite").saveAsTable(f"adb_edaa_dev_ind.silver.{table_name}")
```
